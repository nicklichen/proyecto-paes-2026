## Revisión Directorio Trabajo y Fuentes .CSV

```python
from pathlib import Path

# Validamos que estamos ejecutando codigo en la carpeta que corresponde
DIRECTORIO_ACTUAL = Path.cwd()
print("Directorio de trabajo:")
print(DIRECTORIO_ACTUAL)

# Validamos que el directorio raiz de trabajo contine la carpeta data/raw que es donde hemos dejado los .csv
DIRECTORIO_RAIZ = DIRECTORIO_ACTUAL.parent 
RAW_DIR = DIRECTORIO_RAIZ / "data" / "raw"
print("Directorio Raiz:")
print(DIRECTORIO_RAIZ)

# Declaramos que fuentes esperamos para corroborar
fuentes_esperadas = [
    "PAES-2026-Inscritos-Puntajes",
    "PAES-2026-Matricula",
    "PAES-2026-Oferta-Definitiva-Programas",
    "PAES-2026-Postulantes",
    "PAES-2026-Socioeconomicos"
]
# Para cada fuente esperada, corroboramos que dentro de esta carpeta, exista el .csv
for fuente in fuentes_esperadas:
    ruta = RAW_DIR / fuente
    archivos_csv = list(ruta.glob("*.csv")) if ruta.exists() else []

    if len(archivos_csv) > 0:
        print(f"OK  - {fuente} ({len(archivos_csv)} .csv encontrado)")
    else:
        print(f"ERROR - {fuente} (.csv no encontrado)")
```
